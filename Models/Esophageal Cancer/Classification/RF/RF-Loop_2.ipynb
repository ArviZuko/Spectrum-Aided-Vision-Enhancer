```python
import os
import numpy as np
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt
import tensorflow as tf
```

```python
BATCH_SIZE = 32
IMAGE_SIZE = 640
CHANNELS = 3
```

```python
dataset = tf.keras.preprocessing.image_dataset_from_directory(
    r'/home/smartcity_sp7/EC_Classification/AUG',
    shuffle=True,
    image_size=(IMAGE_SIZE, IMAGE_SIZE),
    batch_size=BATCH_SIZE
)
```

```python
class_names = dataset.class_names
```

```python
def flatten_dataset(dataset):
    flat_data = []
    target_data = []
    for image_batch, labels_batch in dataset:
        for img, label in zip(image_batch, labels_batch):
            img_array = img.numpy().astype('uint8')
            img_flattened = img_array.flatten()
            flat_data.append(img_flattened)
            target_data.append(label.numpy())
    return np.array(flat_data), np.array(target_data)
```

```python
train_size = int(0.85 * len(dataset))
test_size = int(0.15 * len(dataset))
```

```python
train_ds = dataset.take(train_size)
test_ds = dataset.skip(train_size).take(test_size)
```

```python
x_train, y_train = flatten_dataset(train_ds)
x_test, y_test = flatten_dataset(test_ds)
```

```python
best_random_state = None
best_accuracy = 0
```

```python
for random_state in range(25, 50):
    rf = RandomForestClassifier(random_state=random_state)
    rf.fit(x_train, y_train)
    y_pred = rf.predict(x_test)
    accuracy = accuracy_score(y_test, y_pred)
    if accuracy > best_accuracy:
        best_accuracy = accuracy
        best_random_state = random_state
    print(f"Random State: {random_state}, Accuracy: {accuracy:.4f}")
```

```python
print(f"\nBest Random State: {best_random_state}, Best Accuracy: {best_accuracy:.4f}")
```

```python
rf = RandomForestClassifier(random_state=best_random_state)
rf.fit(x_train, y_train)
y_pred = rf.predict(x_test)
accuracy = accuracy_score(y_test, y_pred)
print("The model is {:.2f}% accurate".format(accuracy * 100))
```

```python
report = classification_report(y_test, y_pred, target_names=class_names)
print("Classification Report:\n\n", report)
```

```python
cm = confusion_matrix(y_test, y_pred)
cm_df = pd.DataFrame(cm, columns=class_names, index=class_names)
plt.figure(figsize=(8, 6))
sns.heatmap(cm_df, annot=True, fmt="d", cmap="Blues")
plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.title("Confusion Matrix Heatmap")
plt.show()
```
