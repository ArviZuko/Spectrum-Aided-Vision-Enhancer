```python
import numpy as np
import os
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import xgboost as xgb
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import pandas as pd


train_dir = '/kaggle/input/train-save'
val_dir = '/kaggle/input/valid-save'
test_dir = '/kaggle/input/test-save'


model_path = 'xgboost_SAVEvgg_model.joblib'


base_model = tf.keras.applications.VGG16(weights='imagenet', include_top=False, pooling='avg')


datagen = ImageDataGenerator(rescale=1./255)

train_generator = datagen.flow_from_directory(
    train_dir,
    target_size=(640, 640),
    batch_size=32,
    class_mode='binary', 
    shuffle=False
)

val_generator = datagen.flow_from_directory(
    val_dir,
    target_size=(640, 640),
    batch_size=32,
    class_mode='binary',
    shuffle=False
)

test_generator = datagen.flow_from_directory(
    test_dir,
    target_size=(640, 640),
    batch_size=32,
    class_mode='binary',
    shuffle=False
)


def extract_features(generator):
    features = base_model.predict(generator)
    labels = generator.classes
    return features, labels


train_features, train_labels = extract_features(train_generator)
val_features, val_labels = extract_features(val_generator)
test_features, test_labels = extract_features(test_generator)


xgb_model = xgb.XGBClassifier(use_label_encoder=False, eval_metric='mlogloss')
xgb_model.fit(train_features, train_labels)


val_predictions = xgb_model.predict(val_features)
print("Validation Classification Report:")
print(classification_report(val_labels, val_predictions))


val_cm = confusion_matrix(val_labels, val_predictions)
plt.figure(figsize=(8, 6))
sns.heatmap(val_cm, annot=True, fmt="d", cmap="Blues", xticklabels=train_generator.class_indices.keys(), yticklabels=train_generator.class_indices.keys())
plt.title("Validation Confusion Matrix")
plt.ylabel("Actual")
plt.xlabel("Predicted")
plt.show()


test_predictions = xgb_model.predict(test_features)
print("Test Classification Report:")
print(classification_report(test_labels, test_predictions))


test_cm = confusion_matrix(test_labels, test_predictions)
plt.figure(figsize=(8, 6))
sns.heatmap(test_cm, annot=True, fmt="d", cmap="Greens", xticklabels=test_generator.class_indices.keys(), yticklabels=test_generator.class_indices.keys())
plt.title("Test Confusion Matrix")
plt.ylabel("Actual")
plt.xlabel("Predicted")
plt.show()


joblib.dump(xgb_model, model_path)
print(f"XGBoost model saved to {model_path}")


pred_proba = xgb_model.predict_proba(test_features)
predictions_df = pd.DataFrame({
    'Image': test_generator.filenames,
    'Prediction': test_predictions,
    'Confidence': [max(proba) for proba in pred_proba]
})
predictions_df.to_csv("test_predictions_xgb.csv", index=False)
print("Predictions saved to test_predictions_xgb.csv")
```

```python
import numpy as np
import os
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import xgboost as xgb
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import pandas as pd

# Paths to your test dataset
test_dir = '/kaggle/input/test-rgb'  # Adjust the path as needed
model_path = 'xgboost_RGBvgg_model.joblib'  # Path to your saved XGBoost model

# Load the saved XGBoost model
xgb_model = joblib.load(model_path)

# Load VGG16 as a feature extractor
base_model = tf.keras.applications.VGG16(weights='imagenet', include_top=False, pooling='avg')

# Data generator for test data
datagen = ImageDataGenerator(rescale=1./255)

# Test data generator
test_generator = datagen.flow_from_directory(
    test_dir,
    target_size=(640, 640),
    batch_size=32,
    class_mode='binary',  # or 'categorical' for multi-class
    shuffle=False
)

# Extract features using VGG16 base model
def extract_features(generator):
    features = base_model.predict(generator)
    labels = generator.classes
    return features, labels

# Extract features for the test dataset
test_features, test_labels = extract_features(test_generator)

# Make predictions using the loaded XGBoost model
test_predictions = xgb_model.predict(test_features)

# Get the predicted probabilities (for confidence score)
pred_proba = xgb_model.predict_proba(test_features)

# Create a DataFrame to store image info
predictions_df = pd.DataFrame({
    'Image': test_generator.filenames,
    'True Label': test_labels,
    'Prediction': test_predictions,
    'Confidence': [max(proba) for proba in pred_proba]
})

# Print classification report
print("Test Classification Report:")
print(classification_report(test_labels, test_predictions))

# Confusion Matrix for the test set
test_cm = confusion_matrix(test_labels, test_predictions)
plt.figure(figsize=(8, 6))
sns.heatmap(test_cm, annot=True, fmt="d", cmap="Greens", xticklabels=test_generator.class_indices.keys(), yticklabels=test_generator.class_indices.keys())
plt.title("Test Confusion Matrix")
plt.ylabel("Actual")
plt.xlabel("Predicted")
plt.show()

# Reverse the class_indices mapping
class_labels = {v: k for k, v in test_generator.class_indices.items()}

# Display images in a grid with their true label, predicted label, and confidence level
def plot_images_with_labels(test_generator, predictions_df, num_images=5):
    plt.figure(figsize=(12, 12))
    
    for class_id, class_name in class_labels.items():
        # Filter the predictions for the current class
        class_df = predictions_df[predictions_df['True Label'] == class_id]
        
        # Select 5 images from the current class
        class_images = class_df.head(num_images)
        
        for i, (idx, row) in enumerate(class_images.iterrows()):
            img_path = os.path.join(test_generator.directory, row['Image'])
            img = plt.imread(img_path)
            
            # Get true label, predicted label, and confidence
            true_label = class_labels[row['True Label']]
            predicted_label = class_labels[row['Prediction']]
            confidence = row['Confidence']
            
            # Plot the image with its labels
            plt.subplot(len(class_labels), num_images, i + class_id * num_images + 1)
            plt.imshow(img)
            plt.title(f"True: {true_label}\nPred: {predicted_label}\nConf: {confidence:.2f}")
            plt.axis('off')
    
    plt.tight_layout()
    plt.show()

# Call the function to display 5 images for each class
plot_images_with_labels(test_generator, predictions_df, num_images=5)
```
