```python
import numpy as np
import os
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import pandas as pd

# Paths to train, validation, and test directories
train_dir = '/kaggle/input/train-save'
val_dir = '/kaggle/input/valid-save'
test_dir = '/kaggle/input/test-save'

# Path to save/load model
model_path = 'gbm_SAVEvgg_model.joblib'

# Load VGG16 as a feature extractor
base_model = tf.keras.applications.VGG16(weights='imagenet', include_top=False, pooling='avg')


datagen = ImageDataGenerator(rescale=1./255)

train_generator = datagen.flow_from_directory(
    train_dir,
    target_size=(640, 640),
    batch_size=32,
    class_mode='binary',  # or 'categorical' for multi-class
    shuffle=False
)

val_generator = datagen.flow_from_directory(
    val_dir,
    target_size=(640, 640),
    batch_size=32,
    class_mode='binary',
    shuffle=False
)

test_generator = datagen.flow_from_directory(
    test_dir,
    target_size=(640, 640),
    batch_size=32,
    class_mode='binary',
    shuffle=False
)


def extract_features(generator):
    features = base_model.predict(generator)
    labels = generator.classes
    return features, labels

# Extract features for train, validation, and test datasets
train_features, train_labels = extract_features(train_generator)
val_features, val_labels = extract_features(val_generator)
test_features, test_labels = extract_features(test_generator)

# Create and train the Gradient Boosting Machine model
gbm_model = GradientBoostingClassifier()
gbm_model.fit(train_features, train_labels)

# Evaluate on validation data
val_predictions = gbm_model.predict(val_features)
print("Validation Classification Report:")
print(classification_report(val_labels, val_predictions))


val_cm = confusion_matrix(val_labels, val_predictions)
plt.figure(figsize=(8, 6))
sns.heatmap(val_cm, annot=True, fmt="d", cmap="Blues", xticklabels=train_generator.class_indices.keys(), yticklabels=train_generator.class_indices.keys())
plt.title("Validation Confusion Matrix")
plt.ylabel("Actual")
plt.xlabel("Predicted")
plt.show()

# Evaluate on test data
test_predictions = gbm_model.predict(test_features)
print("Test Classification Report:")
print(classification_report(test_labels, test_predictions))

# Confusion Matrix for test set
test_cm = confusion_matrix(test_labels, test_predictions)
plt.figure(figsize=(8, 6))
sns.heatmap(test_cm, annot=True, fmt="d", cmap="Greens", xticklabels=test_generator.class_indices.keys(), yticklabels=test_generator.class_indices.keys())
plt.title("Test Confusion Matrix")
plt.ylabel("Actual")
plt.xlabel("Predicted")
plt.show()

# Save the model weights
joblib.dump(gbm_model, model_path)
print(f"GBM model saved to {model_path}")


pred_proba = gbm_model.predict_proba(test_features)
predictions_df = pd.DataFrame({
    'Image': test_generator.filenames,
    'Prediction': test_predictions,
    'Confidence': [max(proba) for proba in pred_proba]
})
predictions_df.to_csv("test_predictions_gbm.csv", index=False)
print("Predictions saved to test_predictions_gbm.csv")

import math
import matplotlib.pyplot as plt
import numpy as np

# Load the images from test data generator
test_images, test_labels = next(test_generator)  # Loads one batch of test images and labels
pred_proba = gbm_model.predict_proba(test_features)

# Set up grid size (e.g., 4x4 grid)
grid_size = 4
fig, axes = plt.subplots(grid_size, grid_size, figsize=(15, 15))

for i, ax in enumerate(axes.flat):
    if i >= len(test_images):
        break
    image = test_images[i]
    true_label = test_labels[i]
    pred_label = test_predictions[i]
    confidence = max(pred_proba[i])
    
    # Display the image
    ax.imshow(image)
    ax.axis('off')
    
    # Set title as true label, predicted label, and confidence score
    ax.set_title(f"True: {true_label}, Pred: {pred_label}, Conf: {confidence:.2f}")

plt.tight_layout()
plt.show()
```
