```python
import numpy as np
import os
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import lightgbm as lgb
from sklearn.metrics import classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import pandas as pd

# Paths to train, validation, and test directories
train_dir = '/kaggle/input/save-train'
val_dir = '/kaggle/input/valid-save'
test_dir = '/kaggle/input/test-save'

# Path to save/load model
model_path = 'lightgbm_SAVEvgg19_model.joblib'

# Load VGG19 as a feature extractor
base_model = tf.keras.applications.VGG19(weights='imagenet', include_top=False, pooling='avg')

# Data generator for rescaling
datagen = ImageDataGenerator(rescale=1./255)

# Generators for train, validation, and test sets
train_generator = datagen.flow_from_directory(
    train_dir,
    target_size=(640, 640),
    batch_size=32,
    class_mode='binary',  # or 'categorical' for multi-class
    shuffle=False
)

val_generator = datagen.flow_from_directory(
    val_dir,
    target_size=(640, 640),
    batch_size=32,
    class_mode='binary',
    shuffle=False
)

test_generator = datagen.flow_from_directory(
    test_dir,
    target_size=(640, 640),
    batch_size=32,
    class_mode='binary',
    shuffle=False
)

# Feature extraction function
def extract_features(generator):
    features = base_model.predict(generator)
    labels = generator.classes
    return features, labels

# Extract features for each set
train_features, train_labels = extract_features(train_generator)
val_features, val_labels = extract_features(val_generator)
test_features, test_labels = extract_features(test_generator)

# Train the LightGBM model
lgb_model = lgb.LGBMClassifier()
lgb_model.fit(train_features, train_labels)

# Evaluation on validation data
val_predictions = lgb_model.predict(val_features)
print("Validation Classification Report:")
print(classification_report(val_labels, val_predictions))

val_cm = confusion_matrix(val_labels, val_predictions)
plt.figure(figsize=(8, 6))
sns.heatmap(val_cm, annot=True, fmt="d", cmap="Blues", xticklabels=train_generator.class_indices.keys(), yticklabels=train_generator.class_indices.keys())
plt.title("Validation Confusion Matrix")
plt.ylabel("Actual")
plt.xlabel("Predicted")
plt.show()

# Evaluation on test data
test_predictions = lgb_model.predict(test_features)
print("Test Classification Report:")
print(classification_report(test_labels, test_predictions))

# Confusion Matrix for test set
test_cm = confusion_matrix(test_labels, test_predictions)
plt.figure(figsize=(8, 6))
sns.heatmap(test_cm, annot=True, fmt="d", cmap="Blues", xticklabels=test_generator.class_indices.keys(), yticklabels=test_generator.class_indices.keys())
plt.title("Test Confusion Matrix")
plt.ylabel("Actual")
plt.xlabel("Predicted")
plt.show()

# Save the model
joblib.dump(lgb_model, model_path)
print(f"LightGBM model saved to {model_path}")

# Generate predictions with confidence scores
pred_proba = lgb_model.predict_proba(test_features)
predictions_df = pd.DataFrame({
    'Image': test_generator.filenames,
    'True Label': test_labels,
    'Prediction': test_predictions,
    'Confidence': [max(proba) for proba in pred_proba]
})
predictions_df.to_csv("test_predictions_lgb.csv", index=False)
print("Predictions saved to test_predictions_lgb.csv")

# Map class indices to class names
class_labels = {v: k for k, v in test_generator.class_indices.items()}

# Function to plot images with true label, predicted label, and confidence
def plot_images_with_labels(test_generator, predictions_df, num_images=5):
    plt.figure(figsize=(15, 15))
    classes = list(class_labels.keys())
    
    for class_id in classes:
        # Filter predictions for current class
        class_df = predictions_df[predictions_df['True Label'] == class_id]
        
        # Select a few images from current class
        class_images = class_df.sample(n=min(num_images, len(class_df)))  # Random selection if more images than needed
        
        for i, (idx, row) in enumerate(class_images.iterrows()):
            img_path = os.path.join(test_generator.directory, row['Image'])
            img = plt.imread(img_path)
            
            # Get labels and confidence
            true_label = class_labels[row['True Label']]
            predicted_label = class_labels[row['Prediction']]
            confidence = row['Confidence']
            
            # Plot each image with its label
            plt.subplot(len(classes), num_images, i + class_id * num_images + 1)
            plt.imshow(img)
            plt.title(f"True: {true_label}\nPred: {predicted_label}\nConf: {confidence:.2f}")
            plt.axis('off')
    
    plt.tight_layout()
    plt.show()

# Call the function to display images for each class
plot_images_with_labels(test_generator, predictions_df, num_images=5)
```
