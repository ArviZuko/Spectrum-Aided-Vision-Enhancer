```python
import os

# Define the path to your directory
train = '/content/whitelightaugmnted-1/train/labels'
test = '/content/whitelightaugmnted-1/test/images'
valid = '/content/whitelightaugmnted-1/valid/images'

def count_files(directory):
    try:
        # List all files in the directory
        files = [f for f in os.listdir(directory) if os.path.isfile(os.path.join(directory, f))]
        return len(files)
    except FileNotFoundError:
        print("Directory not found.")
        return 0

# Get the number of files
print(f"Number of files in train: {count_files(train)}")
print(f"Number of files in test: {count_files(test)}")
print(f"Number of files in valid: {count_files(valid)}")
```

```python
import cv2
from ultralytics import YOLO
import os
from pathlib import Path

# Load the YOLOv10 model with your trained weights
model = YOLO('/content/runs/detect/train2/weights/best.pt')

# Path to the test images folder
test_folder = '/content/whitelightaugmnted-1/test/images'

# Create an output folder for saving the images with bounding boxes
output_folder = '/content/runs/check'
os.makedirs(output_folder, exist_ok=True)

# Loop through all images in the test folder
for img_file in Path(test_folder).glob('*.jpg'):  # Adjust file extension if necessary
    img_path = str(img_file)

    # Run inference
    results = model(img_path)

    # Extract image and bounding boxes
    result_img = results[0].plot()  # Draw the bounding boxes on the image

    # Save or display the image with bounding boxes
    output_path = os.path.join(output_folder, img_file.name)
    cv2.imwrite(output_path, result_img)

    # Print the image file name (now ensuring it's a string)
    print(f"Processed {img_file.name}")
```

```python
import shutil
import os

def zip_folder(folder_path, output_path):
    """
    Zips the entire folder specified by folder_path into a zip file at output_path.

    Args:
    folder_path (str): Path to the folder to be zipped.
    output_path (str): Path to the output zip file.
    """
    # Ensure the folder path exists
    if not os.path.isdir(folder_path):
        raise FileNotFoundError(f"The folder {folder_path} does not exist")

    # Create a zip archive of the folder
    shutil.make_archive(output_path, 'zip', folder_path)

# Example usage
folder_path = '/content/working/runs'  # Replace with the path to your folder
output_path = '/content/working/WLI'  # Replace with the path where you want the zip file to be saved
zip_folder(folder_path, output_path)
```
