```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
std24 = np.load('data_cap1.npy')
```

```python
tar_inflate = 16
tar_size = tar_inflate * 2
def get_region(img, y, x, inflate):
    return img[y-inflate:y+inflate, x-inflate:x+inflate, :]
```

```python
tar_std = np.zeros((tar_inflate*2*4, tar_inflate*2*6, 3), dtype = 'uint8')

for i in range(24):
    x_id = int(i % 6)
    y_id = int(i / 6)
    
    x_start = x_id * tar_size
    y_start = y_id * tar_size
    x_end = x_start + tar_size
    y_end = y_start + tar_size
    
    c = std24[i]
    tar_std[y_start:y_end, x_start:x_end,:] = c[::-1]
```

```python
plt.figure(figsize=(6, 6))

plt.title('std24')
plt.axis('off')
plt.imshow(cv2.cvtColor(tar_std, cv2.COLOR_BGR2RGB))

plt.show()
```

```python
cv2.imwrite('cap1.bmp', tar_std)
```
